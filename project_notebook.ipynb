**Data Visualization**

For the data visualization a chloropleth map will be made using plotly. To make sure the colours are divided by a range of mortalities per 100.000 people, we need to set the ranges. To determine a suitable range the minimum and maximum values are important.

```python
print(df_combined["Data_Value"].min())
print(df_combined["Data_Value"].max())
```

To set appropriated ranges, the spread of values is important. To check this a boxplot will be used.

While making the boxplot I discovered that the data type of the column Data_values in (partly) df_combined is string. Some values contain a comma and others a dot. Before making the boxplot the datatype will be set to float and the commas will be replaced with dots.

```python
df_combined["Data_Value"] = df_combined["Data_Value"].str.replace(",", ".")
df_combined["Data_Value"] = df_combined["Data_Value"].astype(float)
```

```python
fig = px.box(
    df_combined,
    y="Data_Value",
    title="Distribution of Heart Disease Mortality Rates"
)

fig.show()
```

The q1 is at 258.1 and q3 is at 383.6 so the chosen ranges will be 50 mortalities per 100.000 people long to make sure the distribution is also visible within that range.

```python
def mortality_range(value):
    if 150 <= value < 200:
        return "150-200"
    elif 200 <= value < 250:
        return "200-250"
    elif 250 <= value < 300:
        return "250-300"
    elif 300 <= value < 350:
        return "300-350"
    elif 350 <= value < 400:
        return "350-400"
    elif 400 <= value < 450:
        return "400-450"
    elif 450 <= value < 500:
        return "450-500"
    elif 500 <= value < 550:
        return "500-550"
    elif 550 <= value < 600:
        return "550-600"
    else:
        return "Outside range"

df_combined["Mortality_Range"] = df_combined["Data_Value"].map(mortality_range)

print("Number of values that is not within one of the ranges chosen:")
print(sum(df_combined["Mortality_Range"] == "Outside range"))
```

The determine the influence of gender on the mortality rates, the different values in sex will be seperated into different dataframes.

```python
df_overall = df_combined[df_combined["Stratification1"] == "Overall"]
df_male = df_combined[df_combined["Stratification1"] == "Male"]
df_female = df_combined[df_combined["Stratification1"] == "Female"]
```

```python
# making color maps
# index at the end is needed because a list is return by the function sample_colorscale even if one color is asked
color_map = {
    "150-200": px.colors.sample_colorscale("Reds", [0])[0],
    "200-250": px.colors.sample_colorscale("Reds", [0.125])[0],
    "250-300": px.colors.sample_colorscale("Reds", [0.25])[0],
    "300-350": px.colors.sample_colorscale("Reds", [0.375])[0],
    "350-400": px.colors.sample_colorscale("Reds", [0.5])[0],
    "400-450": px.colors.sample_colorscale("Reds", [0.625])[0],
    "450-500": px.colors.sample_colorscale("Reds", [0.75])[0],
    "500-550": px.colors.sample_colorscale("Reds", [0.875])[0],
    "550-600": px.colors.sample_colorscale("Reds", [1])[0]
}


fig = px.choropleth(df_overall, locations="LocationAbbr", locationmode="USA-states", color="Mortality_Range", scope="usa", animation_frame="Year", hover_name="LocationDesc", hover_data=["Data_Value"], title="Overall Heart Disease Mortality US", color_discrete_map=color_map, category_orders={"Mortality_Range": ["200-250","250-300","300-350","350-400","400-450","450-500","500-550","550-600"]})
fig.show()
```

**Choropleth per sex**

```python
# showing distribution of categories per gender
print(df_male["Mortality_Range"].value_counts().sort_index())
print(df_female["Mortality_Range"].value_counts().sort_index())
```

Using widgets to add dropdown menu to choropleth:

```python
# creating dropdown menu for sex with default "Overall"

sex_dropdown = widgets.Dropdown(
    options=["Male", "Female"],
    value="Female", #feminism
    description="Sex:"
)

# creating new column period to rename slider values

df_combined["Period"] = df_combined["Year"].map({2015: "2014-2016", 2017: "2016-2018", 2019: "2018-2020"})

# creating a function to update the map based on the dropdown menu choice
 
def update_map(sex):

    df_selected = df_combined[df_combined["Stratification1"] == sex]

    fig = px.choropleth(
        df_selected,
        locations="LocationAbbr",
        locationmode="USA-states",
        color="Mortality_Range",
        color_discrete_map=color_map,
        category_orders={"Mortality_Range": ["150-200", "200-250","250-300","300-350","350-400","400-450","450-500","500-550","550-600"]},
        scope="usa",
        animation_frame="Period",
        hover_name="LocationDesc",
        hover_data=["Data_Value"],
        title="Heart Disease Mortality by Sex"
    )

    fig.show()


widgets.interact(update_map, sex=sex_dropdown);
```
